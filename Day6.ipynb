```python
# write the program to find the sum of  factorial series: # doubt
n = int(input("Enter the number: "))
sum_factorials = 0
for i in range(1, n + 1):
    fact = 1
    for j in range(1, i + 1):
        fact *= j
    sum_factorials += fact

print("The sum of factorials is:", sum_factorials)


```

```python
#Make the factorial program using function
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact


print(factorial(7))

```

```python
def sum(*nums):
    result = 0
    for num in nums:
        result += num
    return result
print(sum(1, 2, 3, 4, 5))
print(sum(1, 2, 3))
print(sum(1, 2))
```
